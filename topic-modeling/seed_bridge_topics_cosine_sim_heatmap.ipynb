```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Combine all documents into a single list
all_documents = df1['comb'].tolist() + df2['comb'].tolist() + df3['comb'].tolist() + df4['comb'].tolist() + df5['comb'].tolist() + df6['comb'].tolist()+ df7['comb'].tolist()+ df8['comb'].tolist()+ df9['comb'].tolist()+ df10['comb'].tolist()+ df11['comb'].tolist()+ df12['comb'].tolist()

# Combine all documents into a single corpus
all_corpus = [''.join(doc) for doc in all_documents]

# Initialize a single TfidfVectorizer
vectorizer = TfidfVectorizer()

# Fit the vectorizer to the combined corpus
X_all = vectorizer.fit_transform(all_corpus)

# Transform each dataframe's documents using the fitted vectorizer
X1_dense = vectorizer.transform([''.join(doc) for doc in documents1]).toarray()
X2_dense = vectorizer.transform([''.join(doc) for doc in documents2]).toarray()
X3_dense = vectorizer.transform([''.join(doc) for doc in documents3]).toarray()
X4_dense = vectorizer.transform([''.join(doc) for doc in documents4]).toarray()
X5_dense = vectorizer.transform([''.join(doc) for doc in documents5]).toarray()
X6_dense = vectorizer.transform([''.join(doc) for doc in documents6]).toarray()
X7_dense = vectorizer.transform([''.join(doc) for doc in documents7]).toarray()
X8_dense = vectorizer.transform([''.join(doc) for doc in documents8]).toarray()
X9_dense = vectorizer.transform([''.join(doc) for doc in documents9]).toarray()
X10_dense = vectorizer.transform([''.join(doc) for doc in documents10]).toarray()
X11_dense = vectorizer.transform([''.join(doc) for doc in documents11]).toarray()
X12_dense = vectorizer.transform([''.join(doc) for doc in documents12]).toarray()
```

```python
dfs = [X1_dense, X2_dense, X3_dense, X4_dense, X5_dense, X6_dense, X7_dense, X8_dense, X9_dense, X10_dense, X11_dense, X12_dense]

cos_sim_matrix = []
for i in range(len(dfs)):
    cos_sim_row = []
    for j in range(len(dfs)):
        cos_sim = cosine_similarity(dfs[i], dfs[j])[0][0]
        cos_sim_row.append(cos_sim)
    cos_sim_matrix.append(cos_sim_row)

df_names = ['bluedf', 'pinkdf', 'reddf', 'yellowdf', 'greendf', 'warfakebelgorod_g_p',"zarussia_1_g_pr","ua_tribunal_r_gpb","voin_dv_g_prb", "Lunay14_g_prb","orchestra_w_g_prb","rustroyka1945_g_prb"]
plt.figure(figsize=(12, 9))
sns.heatmap(cos_sim_matrix, annot=True, fmt='.4f', xticklabels=df_names, yticklabels=df_names, cmap='Blues')
plt.title('Cosine Similarity Heatmap between Dataframes')
plt.xlabel('Dataframes')
plt.ylabel('Dataframes')
```
